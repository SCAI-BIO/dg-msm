```python
from pathlib import Path
from experiments.config import MetaConfig

dataset_name = "commpass"

dataset_path = f"data/{dataset_name}_meta.json"


experiment_name = f"{dataset_name}_crossval"

meta_cfg = MetaConfig(
    dataset_path    = dataset_path,
    experiment_name = experiment_name,
    seed            = 42,
    n_splits        = 5,
    n_trials        = 100,
    val_size        = 0.2,
    save_dir        = "results",
)

meta_cfg = meta_cfg.with_updates(save_dir=f"results/{dataset_name}")

configs_dir = Path("configs")
configs_dir.mkdir(parents=True, exist_ok=True)
config_filename = f"{experiment_name}.yaml"
config_path = configs_dir / config_filename

meta_cfg.to_yaml(config_path)

print("Run experiment with: \n")
print(f"tmux new -s experiment python run_exp.py --config {config_path}")

```
